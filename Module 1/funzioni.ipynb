```python
print(any([False, False, True]))       # True
print(any([False, False, False]))      # False
print(any(x > 5 for x in [1, 3, 8]))   # True: c'è almeno un x maggiore di 5
print(any(x > 5 for x in [1, 3, 4]))   # False
```

```python
T = {(1, 2), (2, 3)}
print(any(a == 3 for a, b in T))
```

```python
def almeno_una(R, A):
    return all( any(a == x for a, b in R) for x in A)

A = {1, 2, 3}
print(almeno_una({(1, 2), (2, 3)}, A))
print(almeno_una({(1, 2), (2, 3), (3, 1)}, A))
print(almeno_una({(1, 2), (1, 3), (2, 3)}, A))
```

```python
def al_piu_una(R):
    return all( b == d for a, b in R for c, d in R if a == c)


print(al_piu_una({(1, 2), (2, 3), (3, 1)}))
print(al_piu_una({(1, 2), (1, 3), (2, 3)}))
```

```python
def funzione(R, A):
    return almeno_una(R, A) and al_piu_una(R)

A = {1, 2, 3}
P = {(1, 2), (2, 3), (3, 1)}
Q = {(1, 2), (1, 3), (2, 3)}
T = {(1, 2), (2, 3)}

print(funzione(P, A))
print(funzione(Q, A))
print(funzione(T, A))
```

```python
def iniettiva(R):
    return all(a == c for a, b in R for c, d in R if b == d)

f1 = {(1, "a"), (2, "b"), (3, "c")}
g = {(1, "a"), (2, "b"), (3, "a")}
print(iniettiva(f1))   # cosa ti aspetti?
print(iniettiva(g))    # cosa ti aspetti?
```

```python
def suriettiva(R, B):
    return all(any(b == y for a, b in R) for y in B)

B = {"a", "b", "c"}
print(suriettiva(f1, B))   # cosa ti aspetti?
print(suriettiva(g, B))    # cosa ti aspetti?
```

```python
def biiettiva(R, A, B):
    return funzione(R, A) and iniettiva(R) and suriettiva(R, B)

A = {1, 2, 3}
B = {"a", "b", "c"}
print(biiettiva(f1, A, B))   # cosa ti aspetti?
print(biiettiva(g, A, B))    # cosa ti aspetti?
```
