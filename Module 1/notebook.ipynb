```python
def relazione(A, condizione):
    return {(x, y) for x in A for y in A if condizione(x, y)}


def divide(x, y):
    return y % x == 0


def simmetrica(R):
    return all((b, a) in R for a, b in R)

def minore(x, y):
    return x < y


def divide(x, y):
    return y % x == 0


def somma_pari(x, y):
    return (x + y) % 2 == 0

def riflessiva(R, A):
    return all((x, x) in R for x in A)
```

```python
A = {1, 2, 3, 4, 5, 6}
D = relazione(A, divide)
M = relazione(A, minore)
S = relazione(A, somma_pari)

print(simmetrica(D))
print(simmetrica(M))
print(simmetrica(S))
```

```python
R = {(1, 2), (2, 3), (3, 3)}

for a, b in R:
    for c, d in R:
        if b == c:
            print((a, b), "poi", (c, d))
```

```python
R = {(1, 2), (2, 3), (3, 3)}

for a, b in R:
    for c, d in R:
        if b == c:
            print((a, b), "poi", (c, d), "-> richiesta", (a, d), ":", (a, d) in R)
```

```python
def transitiva(R):
    return all((a, d) in R for a, b in R for c, d in R if b == c)

print(transitiva({(1, 2), (2, 3), (3, 3)}))            # cosa ti aspetti?
print(transitiva({(1, 2), (2, 3), (1, 3), (3, 3)}))    # cosa ti aspetti?
```

```python
def equivalenza(R, A):
    return riflessiva(R, A) and simmetrica(R) and transitiva(R)

A = {1, 2, 3, 4, 5, 6}
D = relazione(A, divide)
M = relazione(A, minore)
S = relazione(A, somma_pari)

print(equivalenza(D, A))
print(equivalenza(M, A))
print(equivalenza(S, A))
```

```python
for nome, rel in [("D", D), ("M", M), ("S", S)]:
    print(nome, riflessiva(rel, A), simmetrica(rel), transitiva(rel))
```
