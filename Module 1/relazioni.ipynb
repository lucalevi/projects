```python
"""
Fase 1 - Modulo 1.1: insiemi, relazioni, funzioni
Relazioni su un insieme finito, rappresentate come insiemi di coppie (tuple).
"""


def relazione(A, condizione):
    """Insieme delle coppie (x, y) di A x A per cui condizione(x, y) e' vera."""
    return {(x, y) for x in A for y in A if condizione(x, y)}


def riflessiva(R, A):
    """Per ogni x in A, la coppia (x, x) sta in R."""
    return all((x, x) in R for x in A)


def simmetrica(R):
    """Per ogni (a, b) in R, anche la coppia (b, a) sta in R."""
    return all((b, a) in R for a, b in R)


def transitiva(R):
    """Per ogni catena (a, b), (b, d) in R, anche (a, d) sta in R."""
    return all((a, d) in R for a, b in R for c, d in R if b == c)


def equivalenza(R, A):
    """R e' una relazione di equivalenza: riflessiva, simmetrica e transitiva."""
    return riflessiva(R, A) and simmetrica(R) and transitiva(R)


# --- Condizioni di esempio ---------------------------------------------------

def minore(x, y):
    return x < y


def divide(x, y):
    return y % x == 0


def somma_pari(x, y):
    return (x + y) % 2 == 0


if __name__ == "__main__":
    A = {1, 2, 3, 4, 5, 6}
    D = relazione(A, divide)
    M = relazione(A, minore)
    S = relazione(A, somma_pari)

    print("nome  riflessiva  simmetrica  transitiva  equivalenza")
    for nome, rel in [("D", D), ("M", M), ("S", S)]:
        print(
            nome,
            "    ",
            riflessiva(rel, A),
            "     ",
            simmetrica(rel),
            "     ",
            transitiva(rel),
            "     ",
            equivalenza(rel, A),
        )
```
